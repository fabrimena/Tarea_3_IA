## 4 — Dataset MVTec (código + verificación)

```python
from pathlib import Path
from typing import List, Optional

import torch
from PIL import Image as PILImage
from pytorch_lightning import LightningDataModule
from torch.utils.data import DataLoader, Dataset, random_split
from torchvision import transforms


def _list_images(folder: Path) -> List[Path]:
    if not folder.exists():
        return []
    exts = {".png", ".jpg", ".jpeg", ".bmp"}
    return sorted(p for p in folder.rglob("*") if p.suffix.lower() in exts)


def _resolve_val_folder(cls_root: Path, val_folder_names: List[str]) -> Optional[Path]:
    for name in val_folder_names:
        folder = cls_root / name
        if folder.is_dir() and _list_images(folder):
            return folder
    return None


def _has_test_good(root: Path, classes: List[str]) -> bool:
    for cls in classes:
        good_dir = root / cls / "test" / "good"
        if good_dir.is_dir() and _list_images(good_dir):
            return True
    return False


class MVTecImageDataset(Dataset):
    """Carga imágenes MVTec AD: train/good, validation o test/good, test completo."""

    def __init__(
        self,
        root: str,
        classes: List[str],
        image_size: int = 128,
        split: str = "train",
        val_folder_names: Optional[List[str]] = None,
        val_source: str = "test_good",
        defect_only: bool = False,
        good_only: bool = False,
    ):
        self.root = Path(root)
        self.image_size = image_size
        self.paths: List[Path] = []
        self.labels: List[str] = []
        val_folder_names = val_folder_names or ["validation", "val"]

        for cls in classes:
            cls_root = self.root / cls
            if split == "train":
                folders = [cls_root / "train"]
            elif split == "validation":
                val_dir = _resolve_val_folder(cls_root, val_folder_names)
                if val_dir:
                    folders = [val_dir]
                elif val_source == "test_good":
                    folders = [cls_root / "test" / "good"]
                else:
                    folders = []
            elif split == "test":
                if defect_only:
                    test_root = cls_root / "test"
                    if test_root.exists():
                        for sub in sorted(test_root.iterdir()):
                            if sub.is_dir() and sub.name != "good":
                                self._add_folder(sub, f"{cls}/{sub.name}")
                    continue
                elif good_only:
                    folders = [cls_root / "test" / "good"]
                else:
                    test_root = cls_root / "test"
                    if test_root.exists():
                        for sub in sorted(test_root.iterdir()):
                            if sub.is_dir():
                                label = "good" if sub.name == "good" else f"{cls}/{sub.name}"
                                self._add_folder(sub, label)
                    continue
            else:
                raise ValueError(f"Unknown split: {split}")

            for folder in folders:
                if folder and folder.exists():
                    self._add_folder(folder, f"{cls}/good")

        self.transform = transforms.Compose(
            [
                transforms.Resize((image_size, image_size)),
                transforms.ToTensor(),
            ]
        )

    def _add_folder(self, folder: Path, label: str) -> None:
        for path in _list_images(folder):
            self.paths.append(path)
            self.labels.append(label)

    def __len__(self) -> int:
        return len(self.paths)

    def __getitem__(self, idx: int):
        img = PILImage.open(self.paths[idx]).convert("RGB")
        return self.transform(img), self.labels[idx], str(self.paths[idx])
```

```python
def count_dataset_images(root, classes, val_folder_names=None, val_source="test_good"):
    val_folder_names = val_folder_names or ["validation", "val"]
    root_path = Path(root)
    stats = {}
    use_test_good_val = val_source == "test_good" and _has_test_good(root_path, classes)
    for cls in classes:
        cls_root = root_path / cls
        train_n = len(_list_images(cls_root / "train"))
        val_dir = _resolve_val_folder(cls_root, val_folder_names)
        if val_dir:
            val_n = len(_list_images(val_dir))
            val_src = "validation/"
        elif use_test_good_val:
            val_n = len(_list_images(cls_root / "test" / "good"))
            val_src = "test/good"
        else:
            val_n, val_src = 0, "—"
        test_root = cls_root / "test"
        test_n = len(_list_images(test_root)) if test_root.exists() else 0
        stats[cls] = {"train": train_n, "validation": val_n, "validation_source": val_src, "test": test_n}
    return stats

stats = count_dataset_images(
    str(DATASET_ROOT), list(cfg.data.classes),
    list(cfg.data.val_folder_names), cfg.data.get("val_source", "test_good"),
)
rows = "".join(
    f"<tr><td>{c}</td><td>{s['train']}</td><td>{s['validation']} ({s['validation_source']})</td><td>{s['test']}</td></tr>"
    for c, s in stats.items()
)
display(HTML("<table border=1><tr><th>Clase</th><th>train</th><th>val</th><th>test</th></tr>" + rows + "</table>"))
assert sum(s["train"] for s in stats.values()) > 0
assert sum(s["validation"] for s in stats.values()) > 0
print("Listo para entrenar.")
```

## 6 — Modelos, visualización W&B y entrenamiento

```python
from typing import Tuple

import torch
import torch.nn as nn
from pytorch_lightning import LightningModule



class ConvVAE(nn.Module):
    def __init__(self, in_channels: int = 3, latent_dim: int = 128):
        super().__init__()
        self.latent_dim = latent_dim
        self.enc = nn.Sequential(
            nn.Conv2d(in_channels, 32, 4, 2, 1),
            nn.ReLU(),
            nn.Conv2d(32, 64, 4, 2, 1),
            nn.ReLU(),
            nn.Conv2d(64, 128, 4, 2, 1),
            nn.ReLU(),
            nn.Conv2d(128, 256, 4, 2, 1),
            nn.ReLU(),
            nn.Conv2d(256, 512, 4, 2, 1),
            nn.ReLU(),
        )
        self.fc_mu = nn.Linear(512 * 4 * 4, latent_dim)
        self.fc_logvar = nn.Linear(512 * 4 * 4, latent_dim)
        self.fc_dec = nn.Linear(latent_dim, 512 * 4 * 4)
        self.dec = nn.Sequential(
            nn.ConvTranspose2d(512, 256, 4, 2, 1),
            nn.ReLU(),
            nn.ConvTranspose2d(256, 128, 4, 2, 1),
            nn.ReLU(),
            nn.ConvTranspose2d(128, 64, 4, 2, 1),
            nn.ReLU(),
            nn.ConvTranspose2d(64, 32, 4, 2, 1),
            nn.ReLU(),
            nn.ConvTranspose2d(32, in_channels, 4, 2, 1),
            nn.Sigmoid(),
        )

    def encode(self, x: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor]:
        h = self.enc(x).view(x.size(0), -1)
        return self.fc_mu(h), self.fc_logvar(h)

    def reparameterize(self, mu: torch.Tensor, logvar: torch.Tensor) -> torch.Tensor:
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std

    def decode(self, z: torch.Tensor) -> torch.Tensor:
        h = self.fc_dec(z).view(z.size(0), 512, 4, 4)
        return self.dec(h)

    def forward(self, x: torch.Tensor):
        mu, logvar = self.encode(x)
        z = self.reparameterize(mu, logvar)
        return self.decode(z), mu, logvar, z


class VAELightningModule(LightningModule):
    def __init__(
        self,
        latent_dim: int = 128,
        learning_rate: float = 1e-4,
        kl_weight: float = 0.001,
        loss_type: str = "l1",
        ssim_weight: float = 0.5,
        l1_weight: float = 0.5,
        in_channels: int = 3,
    ):
        super().__init__()
        self.save_hyperparameters()
        self.model = ConvVAE(in_channels=in_channels, latent_dim=latent_dim)
        self.recon_loss = ReconstructionLoss(loss_type, ssim_weight, l1_weight)
        self.kl_weight = kl_weight
        self.learning_rate = learning_rate

    def forward(self, x):
        return self.model(x)

    def _shared_step(self, batch):
        x, _, _ = batch
        recon, mu, logvar, z = self.model(x)
        recon_l = self.recon_loss(recon, x)
        kl = -0.5 * torch.mean(1 + logvar - mu.pow(2) - logvar.exp())
        loss = recon_l + self.kl_weight * kl
        return loss, recon, x, z

    def training_step(self, batch, batch_idx):
        loss, _, _, _ = self._shared_step(batch)
        self.log("train_loss", loss, on_step=True, on_epoch=True, prog_bar=True)
        return loss

    def validation_step(self, batch, batch_idx):
        loss, recon, x, z = self._shared_step(batch)
        self.log("val_loss", loss, on_step=False, on_epoch=True, prog_bar=True)
        return {"loss": loss, "recon": recon, "x": x, "z": z}

    def test_step(self, batch, batch_idx):
        loss, recon, x, z = self._shared_step(batch)
        self.log("test_loss", loss, on_step=False, on_epoch=True)
        return {"loss": loss, "recon": recon, "x": x, "z": z, "labels": batch[1]}

    def configure_optimizers(self):
        return torch.optim.Adam(self.parameters(), lr=self.learning_rate)

    def get_latent(self, x: torch.Tensor) -> torch.Tensor:
        mu, _ = self.model.encode(x)
        return mu


from typing import List

import torch
import torch.nn as nn
from pytorch_lightning import LightningModule



class DoubleConv(nn.Module):
    def __init__(self, in_ch: int, out_ch: int):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1),
            nn.BatchNorm2d(out_ch),
            nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1),
            nn.BatchNorm2d(out_ch),
            nn.ReLU(inplace=True),
        )

    def forward(self, x):
        return self.net(x)


class UNetAutoencoder(nn.Module):
    """U-Net style autoencoder with skip connections for reconstruction."""

    def __init__(self, in_channels: int = 3, bottleneck_channels: int = 128):
        super().__init__()
        self.bottleneck_channels = bottleneck_channels
        self.down1 = DoubleConv(in_channels, 64)
        self.pool1 = nn.MaxPool2d(2)
        self.down2 = DoubleConv(64, 128)
        self.pool2 = nn.MaxPool2d(2)
        self.down3 = DoubleConv(128, 256)
        self.pool3 = nn.MaxPool2d(2)
        self.down4 = DoubleConv(256, 512)
        self.pool4 = nn.MaxPool2d(2)

        self.bottleneck = DoubleConv(512, bottleneck_channels)

        self.up4 = nn.ConvTranspose2d(bottleneck_channels, 512, 2, stride=2)
        self.dec4 = DoubleConv(1024, 512)
        self.up3 = nn.ConvTranspose2d(512, 256, 2, stride=2)
        self.dec3 = DoubleConv(512, 256)
        self.up2 = nn.ConvTranspose2d(256, 128, 2, stride=2)
        self.dec2 = DoubleConv(256, 128)
        self.up1 = nn.ConvTranspose2d(128, 64, 2, stride=2)
        self.dec1 = DoubleConv(128, 64)
        self.out = nn.Conv2d(64, in_channels, 1)

    def forward(self, x: torch.Tensor):
        e1 = self.down1(x)
        e2 = self.down2(self.pool1(e1))
        e3 = self.down3(self.pool2(e2))
        e4 = self.down4(self.pool3(e3))
        b = self.bottleneck(self.pool4(e4))

        d4 = self.up4(b)
        d4 = self.dec4(torch.cat([d4, e4], dim=1))
        d3 = self.up3(d4)
        d3 = self.dec3(torch.cat([d3, e3], dim=1))
        d2 = self.up2(d3)
        d2 = self.dec2(torch.cat([d2, e2], dim=1))
        d1 = self.up1(d2)
        d1 = self.dec1(torch.cat([d1, e1], dim=1))
        return torch.sigmoid(self.out(d1)), b.flatten(1)


class UNetAELightningModule(LightningModule):
    def __init__(
        self,
        bottleneck_channels: int = 128,
        learning_rate: float = 1e-4,
        loss_type: str = "l1",
        ssim_weight: float = 0.5,
        l1_weight: float = 0.5,
        in_channels: int = 3,
    ):
        super().__init__()
        self.save_hyperparameters()
        self.model = UNetAutoencoder(in_channels, bottleneck_channels)
        self.recon_loss = ReconstructionLoss(loss_type, ssim_weight, l1_weight)
        self.learning_rate = learning_rate

    def forward(self, x):
        return self.model(x)

    def _shared_step(self, batch):
        x, _, _ = batch
        recon, latent = self.model(x)
        loss = self.recon_loss(recon, x)
        return loss, recon, x, latent

    def training_step(self, batch, batch_idx):
        loss, _, _, _ = self._shared_step(batch)
        self.log("train_loss", loss, on_step=True, on_epoch=True, prog_bar=True)
        return loss

    def validation_step(self, batch, batch_idx):
        loss, recon, x, z = self._shared_step(batch)
        self.log("val_loss", loss, on_step=False, on_epoch=True, prog_bar=True)
        return {"loss": loss, "recon": recon, "x": x, "z": z}

    def test_step(self, batch, batch_idx):
        loss, recon, x, z = self._shared_step(batch)
        self.log("test_loss", loss, on_step=False, on_epoch=True)
        return {"loss": loss, "recon": recon, "x": x, "z": z, "labels": batch[1]}

    def configure_optimizers(self):
        return torch.optim.Adam(self.parameters(), lr=self.learning_rate)

    def get_latent(self, x: torch.Tensor) -> torch.Tensor:
        _, z = self.model(x)
        return z


from typing import Any, Dict, List, Optional

import numpy as np
import torch
import wandb
from pytorch_lightning import Callback, LightningModule, Trainer
from pytorch_lightning.utilities import rank_zero_only
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE


def _tensor_to_wandb_images(orig: torch.Tensor, recon: torch.Tensor, max_n: int = 16):
    n = min(max_n, orig.size(0))
    panels = []
    for i in range(n):
        o = (orig[i].detach().cpu().numpy().transpose(1, 2, 0) * 255).astype(np.uint8)
        r = (recon[i].detach().cpu().numpy().transpose(1, 2, 0) * 255).astype(np.uint8)
        panel = np.concatenate([o, r], axis=1)
        panels.append(wandb.Image(panel, caption=f"original|recon #{i}"))
    return panels


@rank_zero_only
def _log_wandb(trainer: Trainer, key: str, value: Any, step: Optional[int] = None):
    if trainer.logger is None:
        return
    experiment = getattr(trainer.logger, "experiment", None)
    if experiment is not None and hasattr(experiment, "log"):
        experiment.log({key: value}, step=step)


class WandbVisualizationCallback(Callback):
    """Logs reconstructions, t-SNE latent space, and test anomaly panels to W&B."""

    def __init__(
        self,
        num_images: int = 16,
        num_latent_samples: int = 200,
        log_tsne: bool = True,
        log_pca: bool = True,
    ):
        self.num_images = num_images
        self.num_latent_samples = num_latent_samples
        self.log_tsne = log_tsne
        self.log_pca = log_pca
        self._val_outputs: List[Dict] = []
        self._loss_rows: List[List] = []


    def on_train_epoch_end(self, trainer: Trainer, pl_module: LightningModule) -> None:
        if trainer.sanity_checking:
            return
        loss = trainer.callback_metrics.get("train_loss")
        if loss is not None:
            self._loss_rows.append([int(trainer.current_epoch), "train", float(loss)])

    def on_validation_epoch_end(self, trainer: Trainer, pl_module: LightningModule) -> None:
        if trainer.sanity_checking:
            return
        loader = trainer.datamodule.val_dataloader()
        pl_module.eval()
        device = pl_module.device
        origs, recons, latents = [], [], []
        with torch.no_grad():
            for batch in loader:
                x, _, _ = batch
                x = x.to(device)
                if hasattr(pl_module, "model") and hasattr(pl_module.model, "encode"):
                    recon, mu, logvar, z = pl_module.model(x)
                    latent = mu
                else:
                    recon, latent = pl_module.model(x)
                origs.append(x.cpu())
                recons.append(recon.cpu())
                latents.append(latent.cpu())
                n_img = sum(t.size(0) for t in origs)
                n_lat = sum(t.size(0) for t in latents)
                if n_img >= self.num_images and n_lat >= self.num_latent_samples:
                    break
        orig = torch.cat(origs)[: self.num_images]
        recon = torch.cat(recons)[: self.num_images]
        latent = torch.cat(latents)[: self.num_latent_samples]

        panels = _tensor_to_wandb_images(orig, recon, self.num_images)
        step = trainer.global_step
        _log_wandb(trainer, "val/reconstructions", panels, step=step)

        if latent.size(0) >= 4:
            z_np = latent.numpy()
            if self.log_pca and z_np.shape[1] >= 2:
                pca = PCA(n_components=2, random_state=42)
                emb_pca = pca.fit_transform(z_np)
                table_pca = wandb.Table(data=emb_pca.tolist(), columns=["pc1", "pc2"])
                _log_wandb(
                    trainer,
                    "val/latent_pca",
                    wandb.plot.scatter(
                        table_pca,
                        "pc1",
                        "pc2",
                        title=f"PCA latente epoch {trainer.current_epoch} (n={z_np.shape[0]})",
                    ),
                    step=step,
                )
            if self.log_tsne:
                perplexity = min(30, max(2, z_np.shape[0] - 1))
                emb = TSNE(
                    n_components=2,
                    perplexity=perplexity,
                    random_state=42,
                    init="pca",
                    learning_rate="auto",
                ).fit_transform(z_np)
                table = wandb.Table(data=emb.tolist(), columns=["x", "y"])
                _log_wandb(
                    trainer,
                    "val/latent_tsne",
                    wandb.plot.scatter(
                        table,
                        "x",
                        "y",
                        title=f"t-SNE latente epoch {trainer.current_epoch} (n={z_np.shape[0]})",
                    ),
                    step=step,
                )


        if self._loss_rows or trainer.callback_metrics.get("val_loss") is not None:
            rows = list(self._loss_rows)
            val_loss = trainer.callback_metrics.get("val_loss")
            if val_loss is not None:
                rows.append([int(trainer.current_epoch), "val", float(val_loss)])
            if rows:
                table = wandb.Table(data=rows, columns=["epoch", "split", "loss"])
                _log_wandb(
                    trainer,
                    "charts/train_val_loss",
                    wandb.plot.line(
                        table,
                        "epoch",
                        "loss",
                        stroke="split",
                        title=f"Train vs Val — epoch {trainer.current_epoch}",
                    ),
                    step=step,
                )

    def on_test_epoch_end(self, trainer: Trainer, pl_module: LightningModule) -> None:
        loader = trainer.datamodule.test_dataloader()
        pl_module.eval()
        device = pl_module.device
        good_orig, good_recon = [], []
        defect_orig, defect_recon = [], []

        with torch.no_grad():
            for batch in loader:
                x, labels, _ = batch
                x = x.to(device)
                if hasattr(pl_module, "model") and hasattr(pl_module.model, "encode"):
                    recon, _, _, _ = pl_module.model(x)
                else:
                    recon, _ = pl_module.model(x)
                for i, lbl in enumerate(labels):
                    is_good = lbl.endswith("/good") or lbl == "good"
                    t_o, t_r = x[i : i + 1].cpu(), recon[i : i + 1].cpu()
                    if is_good and len(good_orig) < self.num_images:
                        good_orig.append(t_o)
                        good_recon.append(t_r)
                    elif not is_good and len(defect_orig) < self.num_images:
                        defect_orig.append(t_o)
                        defect_recon.append(t_r)

        step = trainer.global_step
        if good_orig:
            o = torch.cat(good_orig)[: self.num_images]
            r = torch.cat(good_recon)[: self.num_images]
            _log_wandb(trainer, "test/reconstructions_good", _tensor_to_wandb_images(o, r), step=step)
        if defect_orig:
            o = torch.cat(defect_orig)[: self.num_images]
            r = torch.cat(defect_recon)[: self.num_images]
            _log_wandb(trainer, "test/reconstructions_defect", _tensor_to_wandb_images(o, r), step=step)

        self._log_reconstruction_histograms(trainer, pl_module, loader, device)

    def _log_reconstruction_histograms(
        self, trainer: Trainer, pl_module: LightningModule, loader, device
    ) -> None:
        errors_good: List[float] = []
        errors_defect: Dict[str, List[float]] = {}

        with torch.no_grad():
            for batch in loader:
                x, labels, _ = batch
                x = x.to(device)
                if hasattr(pl_module, "model") and hasattr(pl_module.model, "encode"):
                    recon, _, _, _ = pl_module.model(x)
                else:
                    recon, _ = pl_module.model(x)
                err = torch.mean((recon - x) ** 2, dim=(1, 2, 3)).cpu().numpy()
                for e, lbl in zip(err, labels):
                    if lbl.endswith("/good") or lbl == "good":
                        errors_good.append(float(e))
                    else:
                        errors_defect.setdefault(lbl, []).append(float(e))

        hist_data = []
        for v in errors_good:
            hist_data.append(["good", v])
        for defect_class, vals in errors_defect.items():
            for v in vals:
                hist_data.append([defect_class, v])

        if hist_data:
            table = wandb.Table(data=hist_data, columns=["class", "reconstruction_mse"])
            _log_wandb(
                trainer,
                "test/reconstruction_error_histogram",
                wandb.plot.histogram(
                    table,
                    "reconstruction_mse",
                    title="Reconstruction error by class (test)",
                ),
                step=trainer.global_step,
            )
```

```python
"""Rutina de entrenamiento reutilizable desde Hydra CLI o Jupyter."""

from pathlib import Path
from typing import Optional

import pytorch_lightning as pl
import torch
from omegaconf import DictConfig, OmegaConf
from pytorch_lightning.loggers import WandbLogger



def build_model(cfg: DictConfig):
    loss_cfg = cfg.loss
    data_cfg = cfg.data
    if cfg.model.name == "vae":
        return VAELightningModule(
            latent_dim=cfg.model.latent_dim,
            learning_rate=cfg.model.learning_rate,
            kl_weight=cfg.model.kl_weight,
            loss_type=loss_cfg.type,
            ssim_weight=loss_cfg.get("ssim_weight", 0.5),
            l1_weight=loss_cfg.get("l1_weight", 0.5),
            in_channels=data_cfg.num_channels,
        )
    if cfg.model.name == "unet":
        return UNetAELightningModule(
            bottleneck_channels=cfg.model.bottleneck_channels,
            learning_rate=cfg.model.learning_rate,
            loss_type=loss_cfg.type,
            ssim_weight=loss_cfg.get("ssim_weight", 0.5),
            l1_weight=loss_cfg.get("l1_weight", 0.5),
            in_channels=data_cfg.num_channels,
        )
    raise ValueError(f"Modelo desconocido: {cfg.model.name}")


def build_datamodule(cfg: DictConfig) -> MVTecDataModule:
    return MVTecDataModule(
        root=cfg.data.root,
        classes=list(cfg.data.classes),
        image_size=cfg.data.image_size,
        batch_size=cfg.data.batch_size,
        num_workers=cfg.data.num_workers,
        val_split=cfg.data.val_split,
        use_predefined_splits=cfg.data.get("use_predefined_splits", True),
        val_folder_names=list(cfg.data.get("val_folder_names", ["validation", "val"])),
        val_source=cfg.data.get("val_source", "test_good"),
    )


def run_training(cfg: DictConfig, output_dir: Optional[Path] = None) -> str:
    """
    Entrena y evalúa con PyTorch Lightning + W&B.
    Devuelve el run id de W&B.
    """
    import os

    pl.seed_everything(cfg.seed, workers=True)
    from pathlib import Path as _Path

    root = _Path(cfg.data.root).parent if hasattr(cfg.data, "root") else _Path.cwd()
    wb_project, wb_entity = resolve_wandb_settings(
        project=os.environ.get("WANDB_PROJECT") or cfg.logger.project,
        entity=cfg.logger.entity,
        project_root=root,
    )

    if cfg.trainer.accelerator == "gpu" and not torch.cuda.is_available():
        raise RuntimeError(
            "Se configuró accelerator=gpu pero CUDA no está disponible. "
            "Instale PyTorch con soporte CUDA."
        )

    datamodule = build_datamodule(cfg)
    model = build_model(cfg)

    wandb_logger = WandbLogger(
        project=wb_project,
        entity=wb_entity,
        name=cfg.experiment_name,
        tags=list(cfg.logger.tags),
        save_dir=cfg.logger.save_dir,
        log_model=cfg.logger.log_model,
        config=OmegaConf.to_container(cfg, resolve=True),
    )

    trainer = pl.Trainer(
        max_epochs=cfg.trainer.max_epochs,
        accelerator=cfg.trainer.accelerator,
        devices=cfg.trainer.devices,
        precision=cfg.trainer.precision,
        gradient_clip_val=cfg.trainer.gradient_clip_val,
        log_every_n_steps=cfg.trainer.log_every_n_steps,
        check_val_every_n_epoch=cfg.trainer.check_val_every_n_epoch,
        enable_progress_bar=cfg.trainer.enable_progress_bar,
        logger=wandb_logger,
        callbacks=[WandbVisualizationCallback(num_images=16, num_latent_samples=200)],
    )

    try:
        trainer.fit(model, datamodule=datamodule)
        trainer.test(model, datamodule=datamodule)
        run_id = wandb_logger.experiment.id
    finally:
        # Cierra el run activo para que el siguiente experimento cree uno nuevo.
        import wandb

        if wandb.run is not None:
            wandb.finish()

    if output_dir is not None:
        output_dir.mkdir(parents=True, exist_ok=True)
        (output_dir / "wandb_run_id.txt").write_text(run_id, encoding="utf-8")
    return run_id



def save_completed_runs(path: Path, completed: Dict[str, str]) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(json.dumps(completed, indent=2), encoding="utf-8")


def load_completed_runs(path: Path) -> Dict[str, str]:
    if not path.exists():
        return {}
    return json.loads(path.read_text(encoding="utf-8"))
"""Ejecuta la matriz de experimentos (Hydra) y persiste run ids."""

import json
from pathlib import Path
from typing import Dict, List, Optional

from omegaconf import OmegaConf

import os



def build_experiment_overrides(cfg) -> List[List[str]]:
    return [
        [f"model={m}", f"loss={l}"]
        for m in cfg.experiments.models
        for l in cfg.experiments.losses
    ]


def run_all_experiments(
    load_cfg_fn,
    project_root: Path,
    skip_if_complete: bool = True,
    force_retrain: bool = False,
) -> Dict[str, str]:
    """
    Entrena los 8 experimentos (2 modelos × 4 pérdidas).
    Devuelve {experiment_name: wandb_run_id}.
    """
    cfg_base = load_cfg_fn()
    resolve_wandb_settings(
        project=os.environ.get("WANDB_PROJECT") or cfg_base.logger.project,
        entity=cfg_base.logger.entity,
        project_root=project_root,
        force=force_retrain,
    )
    overrides_list = build_experiment_overrides(cfg_base)
    runs_file = project_root / "wandb_exports" / "completed_runs.json"
    completed = {} if force_retrain else load_completed_runs(runs_file)

    expected = len(overrides_list)
    if skip_if_complete and not force_retrain and len(completed) >= expected:
        print(f"Ya hay {len(completed)}/{expected} runs en {runs_file}.")
        print("Para re-entrenar: FORCE_RETRAIN = True en el notebook.")
        return completed

    pending = [ov for ov in overrides_list if load_cfg_fn(ov).experiment_name not in completed]
    print(f"Entrenamientos pendientes: {len(pending)} de {expected}")

    for ov in overrides_list:
        c = load_cfg_fn(ov)
        name = c.experiment_name
        if skip_if_complete and not force_retrain and name in completed:
            print(f"Omitiendo {name} (ya en completed_runs.json)")
            continue
        print("\n" + "=" * 60)
        print("Entrenando:", name)
        print(OmegaConf.to_yaml(c.trainer))
        run_id = run_training(c)
        completed[name] = run_id
        print("W&B run id:", run_id)
        save_completed_runs(runs_file, completed)

    return completed
```

```python
"""Paneles W&B — descarga de imágenes/tablas desde wandb.ai (sin matplotlib)."""

PANEL_FILE_HINTS = {
    "charts/train_val_loss": ["train_val_loss", "train_val_loss_table"],
    "val/reconstructions": ["val/reconstructions_", "images/val/reconstructions"],
    "test/reconstructions_good": ["reconstructions_good", "test/reconstructions_good"],
    "test/reconstructions_defect": ["reconstructions_defect", "test/reconstructions_defect"],
    "val/latent_tsne": ["latent_tsne_table", "latent_tsne"],
    "val/latent_pca": ["latent_pca_table", "latent_pca"],
    "test/reconstruction_error_histogram": [
        "reconstruction_error_histogram_table",
        "reconstruction_error_histogram",
    ],
}

PANEL_KEYS = list(PANEL_FILE_HINTS.keys())


def download_wandb_table_files(run, key: str, dest: Path) -> list:
    dest.mkdir(parents=True, exist_ok=True)
    hints = PANEL_FILE_HINTS.get(key, [key.replace("/", "_")])
    saved = []
    for f in run.files():
        name = f.name.lower()
        if not name.endswith(".table.json"):
            continue
        if not any(h.lower() in name for h in hints):
            continue
        downloaded = f.download(root=str(dest), replace=True)
        if hasattr(downloaded, 'close'):
            downloaded.close()
        local = dest / f.name
        saved.append(local)
    return saved


def download_wandb_image_files(run, key: str, dest: Path, limit: int = 4) -> list:
    dest.mkdir(parents=True, exist_ok=True)
    hints = PANEL_FILE_HINTS.get(key, [key.replace("/", "_")])
    saved = []
    for f in run.files():
        name = f.name.lower()
        if not name.endswith((".png", ".jpg", ".jpeg", ".webp", ".gif")):
            continue
        if not any(h.lower() in name for h in hints):
            continue
        downloaded = f.download(root=str(dest), replace=True)
        if hasattr(downloaded, 'close'):
            downloaded.close()
        local = dest / f.name
        saved.append(local)
        if len(saved) >= limit:
            break
    return saved


def _render_table_json_as_plotly(json_path: Path, key: str):
    """Renderiza un archivo table.json como gráfica Plotly interactiva."""
    try:
        import plotly.graph_objects as go
        import plotly.express as px
        
        data = json.loads(json_path.read_text(encoding="utf-8"))
        
        # Extraer datos y columnas
        columns = data.get("columns", [])
        table_data = data.get("data", [])
        
        if not columns or not table_data:
            return None
        
        # Para gráficas de líneas (train_val_loss, etc.)
        if "epoch" in columns and ("loss" in columns or "split" in columns):
            df_data = {col: [] for col in columns}
            for row in table_data:
                for i, col in enumerate(columns):
                    df_data[col].append(row[i] if i < len(row) else None)
            
            # Si hay columna "split", usar para agrupar líneas
            if "split" in columns and "loss" in columns:
                fig = px.line(df_data, x=columns[0], y="loss", color="split" if "split" in columns else None,
                             title=f"Gráfica: {key}")
            else:
                fig = px.scatter(df_data, x=columns[0], y=columns[1:], 
                                title=f"Gráfica: {key}")
            fig.update_layout(hovermode="x unified")
            return fig
        
        # Para scatter plots (latent embeddings)
        elif len(columns) >= 2:
            df_data = {col: [] for col in columns}
            for row in table_data:
                for i, col in enumerate(columns):
                    df_data[col].append(row[i] if i < len(row) else None)
            
            if len(columns) == 2:
                fig = px.scatter(df_data, x=columns[0], y=columns[1], 
                                title=f"Gráfica: {key}")
            else:
                fig = px.scatter(df_data, x=columns[0], y=columns[1],
                                title=f"Gráfica: {key}")
            fig.update_layout(hovermode="closest")
            return fig
        
        return None
    except Exception as e:
        print(f"Error renderizando {json_path.name}: {e}")
        return None


def _latest_table_json(search_dirs: list, key: str) -> Optional[Path]:
    hints = PANEL_FILE_HINTS.get(key, [key.replace("/", "_")])
    candidates = []
    for root in search_dirs:
        if not root.exists():
            continue
        for fp in root.rglob("*.table.json"):
            name = fp.as_posix().lower()
            if any(h.lower() in name for h in hints):
                candidates.append(fp)
    if not candidates:
        return None
    return max(candidates, key=lambda p: p.stat().st_mtime)


def _local_panel_search_dirs(project_root: Path, run_id: str) -> list:
    roots = [
        project_root / "wandb_exports" / run_id,
        project_root / "wandb_logs" / "Tarea_3" / run_id,
    ]
    wandb_dir = project_root / "wandb_logs" / "wandb"
    if wandb_dir.exists():
        for d in wandb_dir.glob(f"run-*-{run_id}"):
            roots.append(d / "files" / "media" / "images")
            roots.append(d / "files" / "media")
    return [p for p in roots if p.exists()]


def _find_local_panel_images(project_root: Path, run_id: str, key: str) -> list:
    hints = [h.lower() for h in PANEL_FILE_HINTS.get(key, [key.replace("/", "_")])]
    found = []
    for root in _local_panel_search_dirs(project_root, run_id):
        for fp in root.rglob("*"):
            if fp.suffix.lower() not in {".png", ".jpg", ".jpeg", ".gif", ".webp"}:
                continue
            name = fp.as_posix().lower()
            if any(h in name for h in hints):
                found.append(fp)
    return sorted(found)[:4]


def _panel_search_dirs(media_dir: Path, project_root: Path, run_id: str) -> list:
    return [
        media_dir / run_id,
        *[p for p in _local_panel_search_dirs(project_root, run_id) if p != media_dir / run_id],
    ]


def _wandb_chart_link(run, key: str) -> str:
    anchor = key.replace("/", "_")
    return f"{run.url}?nw=nwuserfabrimena#charts/{anchor}" if run.url else ""


def display_wandb_panel(run, key: str, media_dir: Path, project_root: Path, max_images: int = 2) -> None:
    """Muestra imágenes descargadas de W&B; si no hay PNG, renderiza tablas JSON con Plotly."""
    run_dir = media_dir / run.id / key.replace("/", "_")
    run_dir.mkdir(parents=True, exist_ok=True)
    search_dirs = _panel_search_dirs(media_dir, project_root, run.id)

    # Intentar obtener imágenes PNG
    files = _find_local_panel_images(project_root, run.id, key)
    if not files:
        files = download_wandb_image_files(run, key, run_dir, limit=max_images)
    
    # Mostrar imágenes PNG si existen
    if files:
        display(HTML(f"<h4>{key}</h4>"))
        for fp in files[:max_images]:
            display(Image(filename=str(fp)))
        return

    # Si no hay PNG, intentar renderizar tablas JSON con Plotly
    download_wandb_table_files(run, key, run_dir)
    table_json = _latest_table_json([run_dir], key)
    
    if table_json:
        fig = _render_table_json_as_plotly(table_json, key)
        if fig is not None:
            display(HTML(f"<h4>{key}</h4>"))
            display(fig)
            return
    
    # Si todo falla, mostrar link a W&B
    chart_url = _wandb_chart_link(run, key)
    display(
        HTML(
            f"<h4>{key}</h4>"
            f"<p><i>Gráfica interactiva en W&B.</i> "
            f'<a href="{chart_url or run.url}" target="_blank">Abrir en Weights & Biases →</a></p>'
        )
    )
```

```python
"""Resultados W&B — solo medios descargados de wandb.ai (sin matplotlib)."""

"""Resultados W&B en el notebook — solo medios y tablas descargadas de W&B."""


def get_wandb_api():
    return wandb.Api()


def resolve_entity(entity: Optional[str] = None) -> Optional[str]:
    if entity:
        return str(entity)
    try:
        return get_wandb_api().default_entity
    except Exception:
        return None


def resolve_run_path(project: str, run_id: str, entity: Optional[str] = None) -> str:
    entity = resolve_entity(entity)
    if entity:
        return f"{entity}/{project}/{run_id}"
    return f"{project}/{run_id}"


def _is_missing_project_error(exc: BaseException) -> bool:
    return isinstance(exc, ValueError) and "Could not find project" in str(exc)


def _iter_runs(path: str, per_page: int = 50) -> Iterator:
    api = get_wandb_api()
    runs_iter = api.runs(path, per_page=per_page, order="-created_at")
    while True:
        try:
            yield next(runs_iter)
        except StopIteration:
            break
        except ValueError as exc:
            if _is_missing_project_error(exc):
                break
            raise


def workspace_url(project: str, entity: Optional[str] = None) -> Optional[str]:
    entity = resolve_entity(entity)
    if not entity:
        return None
    return f"https://wandb.ai/{entity}/{project}"


def list_project_runs(
    project: str,
    entity: Optional[str] = None,
    model_name: Optional[str] = None,
    limit: int = 100,
) -> List:
    entity = resolve_entity(entity)
    if not entity:
        return []
    path = f"{entity}/{project}"
    filtered = []
    for run in _iter_runs(path):
        if len(filtered) >= limit:
            break
        if model_name:
            cfg_model = run.config.get("model", {})
            name = (
                cfg_model.get("name")
                if isinstance(cfg_model, dict)
                else run.config.get("model/name")
            )
            if name != model_name:
                continue
        filtered.append(run)
    return filtered


def pick_best_run(runs: List, metric: str = "val_loss") -> Optional[Any]:
    if not runs:
        return None

    def _score(r):
        v = r.summary.get(metric)
        if v is None:
            return float("inf")
        try:
            return float(v)
        except (TypeError, ValueError):
            return float("inf")

    best = min(runs, key=_score)
    if best.summary.get(metric) is None:
        return sorted(runs, key=lambda x: x.created_at, reverse=True)[0]
    return best


def save_completed_runs(path: Path, completed: Dict[str, str]) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(json.dumps(completed, indent=2), encoding="utf-8")


def load_completed_runs(path: Path) -> Dict[str, str]:
    if not path.exists():
        return {}
    return json.loads(path.read_text(encoding="utf-8"))


def display_runs_table(runs: List, title: str) -> None:
    if not runs:
        display(HTML(f"<h3>{title}</h3><p><i>Sin runs en W&B.</i></p>"))
        return
    rows = []
    for r in sorted(runs, key=lambda x: x.name):
        val = r.summary.get("val_loss", "—")
        train = r.summary.get("train_loss", "—")
        loss_name = r.config.get("loss", {}).get("name", "") or r.config.get("loss/name", "")
        rows.append(
            f"<tr><td><a href='{r.url}' target='_blank'>{r.name}</a></td>"
            f"<td>{loss_name}</td><td>{train}</td><td>{val}</td></tr>"
        )
    display(
        HTML(
            f"<h3>{title}</h3><table border='1'>"
            "<tr><th>Run</th><th>Pérdida</th><th>train_loss</th><th>val_loss</th></tr>"
            + "".join(rows)
            + "</table>"
        )
    )


def display_loss_from_wandb(run, media_dir: Path, project_root: Path) -> None:
    """Curva train/val: imágenes exportadas por W&B o enlace al panel Charts."""
    display(
        HTML(
            f'<p><b>Pérdidas (origen W&B):</b> <a href="{run.url}" target="_blank">{run.name}</a> '
            "— pestaña <b>Charts</b> en el run.</p>"
        )
    )
    display_wandb_panel(run, "charts/train_val_loss", media_dir, project_root, max_images=1)
    for pattern in ("train_loss", "val_loss", "loss"):
        extra = media_dir / run.id / "charts_auto" / pattern
        imgs = download_wandb_image_files(run, pattern, extra, limit=2)
        if imgs:
            display(HTML(f"<h4>Chart W&B — {pattern}</h4>"))
            for fp in imgs:
                display(Image(filename=str(fp)))


def display_run_panels(
    run,
    title: str,
    media_dir: Path,
    project_root: Optional[Path] = None,
) -> None:
    root = project_root or media_dir.parent
    display(
        HTML(
            f"<h3>{title}</h3>"
            f"<p><a href='{run.url}' target='_blank'><b>{run.name}</b></a> "
            f"(val_loss={run.summary.get('val_loss', '—')})</p>"
        )
    )
    display_loss_from_wandb(run, media_dir, root)

    for key in PANEL_KEYS:
        if key == "charts/train_val_loss":
            continue
        if key in ("val/latent_tsne", "val/latent_pca"):
            display_wandb_panel(run, key, media_dir, root, max_images=1)
            if key == "val/latent_tsne":
                table_path = _latest_table_json(_panel_search_dirs(media_dir, root, run.id), key)
                n_pts = 0
                if table_path:
                    data = json.loads(table_path.read_text(encoding="utf-8"))
                    n_pts = len(data.get("data", []))
                display(
                    HTML(
                        "<p><b>Espacio latente:</b> el latente es 128D; t-SNE/PCA son "
                        f"<b>proyecciones 2D</b> en W&B ({n_pts} puntos en la última tabla). "
                        "No es el espacio completo. Tras re-entrenar se registran hasta "
                        "<b>200</b> muestras + panel <code>val/latent_pca</code>.</p>"
                    )
                )
            continue
        display_wandb_panel(run, key, media_dir, root, max_images=2)


def display_project_results(
    project: str,
    entity: Optional[str],
    media_dir: Path,
    completed_runs: Optional[Dict[str, str]] = None,
    project_root: Optional[Path] = None,
) -> Dict[str, str]:
    root = project_root or media_dir.parent
    load_dotenv_if_present(root)
    ensure_wandb_login()

    entity = resolve_entity(entity)
    api = get_wandb_api()
    url = workspace_url(project, entity)
    if url:
        display(
            HTML(
                f'<p><b>Workspace W&B:</b> <a href="{url}" target="_blank">{url}</a></p>'
            )
        )

    runs_vae = list_project_runs(project, entity, model_name="vae")
    runs_unet = list_project_runs(project, entity, model_name="unet")
    display_runs_table(runs_vae, "VAE — 4 funciones de pérdida")
    display_runs_table(runs_unet, "U-Net AE — 4 funciones de pérdida")

    if completed_runs:
        display(HTML("<h3>Paneles por experimento (8 runs) — medios W&B</h3>"))
        for exp_name, run_id in sorted(completed_runs.items()):
            try:
                run = api.run(resolve_run_path(project, run_id, entity))
            except Exception as exc:
                display(HTML(f"<p><i>{exp_name} ({run_id}): {exc}</i></p>"))
                continue
            display_run_panels(run, f"Experimento: {exp_name}", media_dir, root)

    best_ids = {}
    for model_name, runs in (("vae", runs_vae), ("unet", runs_unet)):
        best = pick_best_run(runs)
        if best:
            best_ids[model_name] = best.id
            display_run_panels(best, f"Mejor run — {model_name.upper()}", media_dir, root)

    if completed_runs:
        save_completed_runs(media_dir / "completed_runs.json", completed_runs)
    return best_ids
```
